## What This Demonstrates

| Turn | Query | Tools Used | Feature Highlighted |
|------|-------|------------|---------------------|
| 1 | 2^10 + LLM meaning | `calculator` + `lookup` | Multi-tool in one step |
| 2 | Convert 1024m + current time | `unit_converter` + `get_current_time` | Context memory across turns |
| 3 | London weather + Python sum | `get_weather` + `run_python` | Live data + code execution |

```python
def chat_agent():
    history = []
    print("\n" + "=" * 62)
    print("  ReAct Agent  -  Conversational Mode")
    print("  Type 'exit' or 'bye' to end the session.")
    print("=" * 62 + "\n")
    while True:
        query = input("You: ").strip()
        if not query:
            continue
        if query.lower() in ("exit", "quit", "bye", "goodbye"):
            print("\nAgent: Goodbye! It was great chatting with you. Come back anytime!")
            break
        if query.lower() in ("hi", "hello", "hey"):
            print("\nAgent: Hello! I'm your ReAct AI assistant.")
            print("I can help with math, conversions, weather, search, Python code, and more.")
            print("What would you like to explore today?\n")
            continue
        print()
        answer = run_react(query, verbose=True)
        history.append({"role": "user", "content": query})
        history.append({"role": "assistant", "content": answer})
        print(f"\nAgent: {answer}\n")

# Uncomment to run interactively (requires ANTHROPIC_API_KEY and setup cells):
# chat_agent()
```
